### Imports

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer
````

## Question 1
Based on documents = `[ "The plaintiff filed a lawsuit for breach of contract.", "Contract disputes are common in civil litigation.", "The defendant denies all allegations of negligence.", "Personal injury and negligence claims require proof of harm." ]`, create a Python program to assign the documents to two clusters.

````python
documents = [
    "The plaintiff filed a lawsuit for breach of contract.",
    "Contract disputes are common in civil litigation.",
    "The defendant denies all allegations of negligence.",
    "Personal injury and negligence claims require proof of harm."
]
````

````python
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)

kmeans = KMeans(n_clusters=2, random_state=42, n_init='auto')
kmeans.fit(X)

for i, doc in enumerate(documents):
    print(f"Document {i+1} -> Cluster {kmeans.labels_[i]}: \"{doc}\"")
````

## Question 2
How close is document 1 to 2 and document 1 to 3 based on the following contents:
```python
doc1 = "The tenant failed to pay rent on time, violating the lease agreement."
doc2 = "The lease was breached because the monthly rent payment was late."
doc3 = "The company merged with a competitor to dominate the market."
```

````python
doc1 = "The tenant failed to pay rent on time, violating the lease agreement."
doc2 = "The lease was breached because the monthly rent payment was late."
doc3 = "The company merged with a competitor to dominate the market."

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform([doc1, doc2, doc3])

sim_1_2 = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])[0][0]
sim_1_3 = cosine_similarity(tfidf_matrix[0], tfidf_matrix[2])[0][0]

print(f"Cosine similarity (doc1 vs doc2): {sim_1_2:.4f}")
print(f"Cosine similarity (doc1 vs doc3): {sim_1_3:.4f}")
````

## Question 3
Create an LLM code to measure the text similarity between doc1 and doc2, as well as doc1 and doc3.
```python
doc1 = "The tenant failed to pay rent on time, violating the lease agreement."
doc2 = "The lease was breached because the monthly rent payment was late and not sent to the appropriate email."
doc3 = "The company merged with a competitor to dominate the market."
```
Use SentenceTransformer, not OpenAI or LangChain.

````python
doc1 = "The tenant failed to pay rent on time, violating the lease agreement."
doc2 = "The lease was breached because the monthly rent payment was late and not sent to the appropriate email."
doc3 = "The company merged with a competitor to dominate the market."

model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode([doc1, doc2, doc3])

sim_1_2 = cosine_similarity([embeddings[0]], [embeddings[1]])[0][0]
sim_1_3 = cosine_similarity([embeddings[0]], [embeddings[2]])[0][0]

print(f"Semantic similarity (doc1 vs doc2): {sim_1_2:.4f}")
print(f"Semantic similarity (doc1 vs doc3): {sim_1_3:.4f}")
````
