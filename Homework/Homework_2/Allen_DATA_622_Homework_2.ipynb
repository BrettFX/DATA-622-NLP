### Imports

```python
import requests
from bs4 import BeautifulSoup
import string
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer
from nltk.tokenize import word_tokenize
```

### Configurations

```python
# Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

## 1. Read the File
Read the content of https://www.cnn.com/2025/06/13/style/why-luxury-brands-are-so-expensive and print the first 700 characters.

```python
url = "https://www.cnn.com/2025/06/13/style/why-luxury-brands-are-so-expensive"

response = requests.get(url)
print("HTTP Status Code:", response.status_code)

html_content = response.text
print("\nFirst 700 characters of raw HTML:\n")
print(html_content[:700])
```

## 2. Remove HTML Tags

If any HTML tags are present in the file, remove them so that only the raw text remains.

```python
soup = BeautifulSoup(html_content, "html.parser")

# Remove script and style elements
for script in soup(["script", "style"]):
    script.decompose()

clean_text = soup.get_text(separator=" ")
```

```python
# Inspect cleaned text
clean_text
```

## 3. Lower and Remove Punctuation

Convert all text to lowercase and remove all punctuation characters.

```python
# Convert to lowercase
clean_text = clean_text.lower()

# Remove punctuation
clean_text = clean_text.translate(str.maketrans('', '', string.punctuation))
```

```python
# Inspect cleaned text after converting to lowercase and removing punctuation
clean_text
```

## 4. Remove Stopwords

Remove English stopwords from the text. (Use NLTK’s list of stopwords.)

```python
stop_words = set(stopwords.words("english"))
tokens = word_tokenize(clean_text)
filtered_tokens = [word for word in tokens if word not in stop_words]
```

```python
print(f'Total Tokens: {len(tokens)}')
```

```python
# Inspect first 50 filtered tokens
filtered_tokens[:50]
```

## 5. Lemmatize the Text

Lemmatize all remaining words (use NLTK’s *WordNetLemmatizer)* and print the first 50 lemmatized words. Is there any difference in your output if you stemmed the text?

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print("\nFirst 50 Lemmatized Words:\n")
print(lemmatized_words[:50])
```
