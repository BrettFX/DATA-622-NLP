### Step 2. Build a Markov chain based on word transitions.

```python
class MarkovChain:
    def __init__(self, order=2, lowercase=True):
        # Ensure the order parameter is valid
        # The order determines how many previous words define a state
        if order < 1:
            raise ValueError("order must be >= 1")
        self.order = order
        self.lowercase = lowercase
        self.chain = defaultdict(list)
        self.is_fitted = False

    def _preprocess(self, text):
        if self.lowercase:
            text = text.lower()

        # Tokenize the text using a regular expression.
        # Extract words (letters/numbers) and sentence-ending punctuation (. ! ?)
        # Keeping punctuation allows the generator to learn where
        # sentences typically end.
        return re.findall(r"\w+|[.!?]", text.lower())

    def fit(self, text):
        # Convert raw text into tokens
        tokens = self._preprocess(text)

        # Ensure there are enough tokens to create Markov states
        if len(tokens) <= self.order:
            raise ValueError("Training text is too short for the selected order.")

        # Clear any previously learned transitions
        self.chain.clear()

        # Iterate through the tokens and construct states
        # Example with order=2:
        #
        # tokens: [once, upon, a, time]
        #
        # state ("once","upon") -> next word "a"
        # state ("upon","a") -> next word "time"
        for i in range(len(tokens) - self.order):
            # Current state consists of order consecutive tokens
            state = tuple(tokens[i:i + self.order])

            # The word that follows this state in the training corpus
            next_word = tokens[i + self.order]

            # Store the transition in the chain dictionary
            # Multiple next words may exist for the same state,
            # allowing the generator to randomly choose among them
            self.chain[state].append(next_word)

        self.is_fitted = True
        return self

    def generate(self, start_text=None, max_sentences=5):
        # Ensure the model has been trained before attempting generation
        if not self.is_fitted:
            raise RuntimeError("Call fit() before generate().")

        # Preprocess the starting phrase into tokens (words/punctuation)
        tokens = self._preprocess(start_text)
        generated = list(tokens)

        # The Markov "state" is defined by the last order tokens
        # Example: if order=3 and start_text="once upon a time",
        # the state becomes ("upon", "a", "time")
        current_state = tuple(tokens[-self.order:])

        # Track how many sentences have been produced
        sentence_count = 0

        # Generate words until the sentence limit is reached
        while sentence_count < max_sentences:

            # If the current state was never seen during training,
            # the Markov model has no information about what word
            # should come next. EX: If "Once upon a time in a kingdom"
            # doesn't exist, a strict markov chain would stop here.
            #
            # To prevent generation from stopping early, randomly
            # select a valid state from the trained Markov chain.
            # This allows generation to continue even when the
            # starting phrase does not appear in the training text.
            if current_state not in self.chain:
                current_state = random.choice(list(self.chain.keys()))

            # Randomly select the next word based on the observed
            # transitions from the current state in the training data
            next_word = random.choice(self.chain[current_state])

            # Append the generated word to the output sequence
            generated.append(next_word)

            # If punctuation indicates the end of a sentence,
            # increment the sentence counter
            if next_word in [".", "!", "?"]:
                sentence_count += 1

            # Update the Markov state by shifting the window forward.
            # This keeps the last order words as the new state.
            current_state = tuple(generated[-self.order:])

        # Join the generated tokens into a single string
        return " ".join(generated)
```

```python
# Train model using bi-gram
markov_model = MarkovChain(order=2)
markov_model = markov_model.fit(fairy_tale_text)
```

### Step 3. Generate and print a story starting with the provided phrase.

```python
# Generate story
story = markov_model.generate(
    start_text="Once upon a time there was a kingdom",
    max_sentences=10
)

print("Generated Story:")
print(story)
```

**Observation:** Basic Markov chains can only generate text from word sequences observed in training data. To prevent generation from stopping when unseen states occur, a fallback strategy can be used that samples a random known state from the Markov model.

## Exercise 2: LLM-Generated Story
The question is ‘Why Study Data Science Today?’

**Instructions:**

Use an LLM (e.g., OpenAI GPT, Hugging Face Transformers) to generate a creative story about why it is important to study data science today.

### Step 1. Use a summarization or text generation pipeline/model.

```python
# Using Microsoft Phi-3-mini-4k-instruct model for text generation since it is great for instruction-following tasks and can generate coherent text based on the provided prompt.
generator = pipeline(
    "text-generation",
    model="microsoft/Phi-3-mini-4k-instruct",
    device_map="auto"
)
```

### Step 2. Prompt the model to write a story (not just an essay or bullet points).

```python
prompt = """
Write a short imaginative story about a student who discovers why studying data science matters today.
The story should have characters, a problem, and a hopeful ending. Only provide the story and nothing else.
"""

result = generator(
    prompt,
    max_length=2048,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.8
)

story = result[0]["generated_text"]
```

### Step 3. Print the story.

```python
# Ensure the prompt isn't included in the response
if prompt in story:
    idx = story.index(prompt)
    story = story[idx+len(prompt):].strip()
else:
    print('Prompt is not in the story.')
```

```python
print("LLM Generated Story:\n")
print(story)
```

**Observation:** The LLM is pretty quick to generate the story and with a high temperature of `0.8`, the story is always different each time the model is invoked. This quality of the story is clearly significantly better than the story the Markov Chain generated.
